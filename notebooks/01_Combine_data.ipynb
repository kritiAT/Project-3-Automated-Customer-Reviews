## 2. Load the data

```python
RAW_CSV_PATH = "../data/raw/Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products_May19.csv"
RAW_CSV_PATH_2 = "../data/raw/Datafiniti_Amazon_Consumer_Reviews_of_Amazon_Products.csv"
RAW_CSV_PATH_3 = "../data/raw/1429_1.csv"
df = pd.read_csv(RAW_CSV_PATH)
print(df.shape)
df_2 = pd.read_csv(RAW_CSV_PATH_2)
print(df_2.shape)
df_3 = pd.read_csv(RAW_CSV_PATH_3)
print(df_3.shape)
```

```python
df_3[df_3.name.isna() == True].value_counts()
```

```python
imp_cols = ['asins', 'categories', 'id', 'name', 'primaryCategories', 'reviews.rating', 'reviews.text', 'reviews.title']
```

```python
df_comb = pd.concat([df[imp_cols], df_2[imp_cols]]) # , df_3[imp_cols]
print(df_comb.shape)
df_comb.head(10)
```

```python
df_comb.isna().sum()
```

```python
df_comb['dups'] = df_comb.duplicated(['name', 'reviews.rating', 'reviews.text'])
df_comb['dups_2'] = df_comb.duplicated()
df_comb
```

```python
df_comb[df_comb.dups == True]
```

```python
df_comb[['id', 'name']].value_counts()
```

```python
df_comb.groupby('categories').agg(
    cats=('name', set),
    unique_categories=('name', 'nunique')
).sort_values(
    by='unique_categories',
    ascending=False
)
```

```python
df_unique = df_comb[imp_cols].drop_duplicates(subset=['name', 'reviews.rating', 'reviews.text'], ignore_index=True)
df_unique
```

```python
df_unique.shape
```

```python
df_unique.to_csv("../data/processed/combined_data.csv", index=False)
print("Saved:", df_unique.shape)
```
